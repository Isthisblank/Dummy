# How is the distribution of song danceability represented across the dataset?

danceability_% → numerical variable
Mean → compare the average danceability
Median → compare the typical danceability




```python
import pandas as pd
df = pd.read_csv("spotify.csv", encoding = "latin-1")
df
```

```python
df.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

mean = df["danceability_%"].mean()
median = df["danceability_%"].median()

print("Mean:", mean)
print("Median:", median)

sns.set_theme()

sns.displot(data=df, x="danceability_%", kde=True)

plt.axvline(mean, linestyle="--", label=f"Mean: {mean:.1f}")
plt.axvline(median, linestyle=":", label=f"Median: {median:.1f}")

plt.legend()
plt.show()
```

# Conclusion

From this observational study, it can be concluded that the distribution of song danceability is generally centered around the upper-middle range. The mean danceability is approximately 67%, while the median is 69%. Since the mean and median are relatively close to each other, the data appears to be fairly balanced, although there is a slight skew toward lower danceability values.

A noticeable pattern in the data is that most songs have a danceability between approximately 50% and 85%, with the highest concentration of songs being around 70% to 80%. There are fewer songs with very low danceability, especially below 40%, and the number of songs gradually decreases after reaching the highest concentration around 70–80%.
