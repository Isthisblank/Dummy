## Question
Statistically, in what habitat are you most likely to find a poisonous mushroom? this is an observational study in which a data set is looked at and commanalities are observed

## Findings

From this data, we can draw two conclusions:

### 1. By Percentage (Highest Risk)
The environment in which one is most likely to encounter a poisonous mushroom is the **'P' habitat**, where **88%** of all mushrooms are poisonous.

### 2. By Total Count (Largest Population)
However, if we consider the absolute number of poisonous mushrooms (not percentages), the answer would be the **'d' habitat**, as it has the largest total number of both poisonous and non-poisonous mushrooms combined.

## Conclusion
- **Most dangerous habitat** (% risk): 'P' habitat at 88% poisonous
- **Most common habitat** (total count): 'd' habitat with the highest mushroom population
