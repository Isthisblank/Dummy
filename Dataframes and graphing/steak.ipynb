```python
import pandas as pd

df = pd.read_csv("steak.csv")

# Remove the repeated survey header row
df = df.iloc[1:]

# Rename columns
df = df.rename(columns={
    "Have you ever cheated on your significant other?": "Cheated"
})

# Create the table
table = pd.crosstab(
    [df["Education"], df["Age"], df["Gender"]],
    df["Cheated"],
    normalize="index"
) * 100

print(table)
```

```python
table = pd.crosstab(
    df["Gender"],
    df["Cheated"],
    normalize="index"
) * 100

print(table)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))

sns.countplot(data=df, x="Education", hue="Cheated")

plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))

sns.countplot(data=df, x="Gender", hue="Cheated")

plt.xticks(rotation=45, ha="right")
plt.show()
```
